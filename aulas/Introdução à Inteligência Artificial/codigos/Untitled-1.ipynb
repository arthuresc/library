# Regressão Linear

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
# scaled=False: IMC em kg/m² (mesmo R² = 0,32)
X, y = load_diabetes(return_X_y=True,
as_frame=True, scaled=False)
X = X[["bmi"]]
X_tr, X_te, y_tr, y_te = train_test_split(
X, y, random_state=42)
modelo = LinearRegression().fit(X_tr, y_tr)
```

```python
novos = pd.DataFrame({"bmi": [22, 30, 40]})
prev = modelo.predict(novos)
print("Previsões:", prev.round())
print("Inclinação a:", modelo.coef_[0].round(1))
print("Intercepto b:", round(modelo.intercept_, 1))
```
