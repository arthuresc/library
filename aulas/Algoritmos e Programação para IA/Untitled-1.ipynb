```python
def spin_words(sentence):
    palavras = sentence.split(" ")
    textoRetorno = ""
    print(palavras)
    
    for index,palavra in enumerate(palavras):
      print(palavra, "😎")
      if len(palavra) >= 5:
        palavras[index] = palavra[::-1]
    
    for palavra in palavras:
      # textoRetorno = textoRetorno + " " + palavra
      palavras[index] = palavra if len(palavras) == index else palavra + " "
    
    return textoRetorno
  
  
spin_words("Welcome")
```
