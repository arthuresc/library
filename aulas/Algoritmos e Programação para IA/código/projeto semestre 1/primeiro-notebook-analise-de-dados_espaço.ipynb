O valor de ESI (Earth Similarity Index ou Indice de Similaridade com a Terra) é um calculo de média quadratica.

```python
def calcular_esi(row, S_terra=1.0, R_terra=1.0):
  s, r = row["P_FLUX"], row["P_RADIUS"]
  termo_s = ((s - S_terra) / (s + S_terra)) ** 2
  termo_r = ((r - R_terra) / (s + S_terra)) ** 2
  return 1 - np.sqrt(0.5 * (termo_s + termo_r))

df_clean["P_ESI_C"] = df_clean.apply(calcular_esi, axis=1)

# df_clean["P_ESI"].describe()

df_clean["P_ESI_C"].describe()
```
